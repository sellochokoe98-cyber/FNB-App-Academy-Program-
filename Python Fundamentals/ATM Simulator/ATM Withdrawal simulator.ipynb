```python
# Smart ATM Withdrawal Simulator

print("=== Welcome to FNB ATM ===")

# 1. FIXED BANK BALANCE
balance = 500.00

# 2. ASK USER FOR WITHDRAWAL AMOUNT
amount = float(input(f"Your current balance is R{balance:.2f}\nEnter amount to withdraw: R"))

# 3. CHECK CONDITIONS
if amount > 0 and amount <= balance:
    # Enough funds and valid amount
    balance = round(balance - amount, 2)

    print("\n=== TRANSACTION RECEIPT ===")
    print(f"Amount Withdrawn : R{amount:.2f}")
    print(f"Remaining Balance: R{balance:.2f}")
    print("Withdrawal successful!")

elif amount <= 0:
    # Invalid amount
    print("\nTransaction Failed")
    print("Invalid amount. You must withdraw more than R0.00.")

else:
    # Insufficient funds
    print("\nTransaction Failed")
    print("Declined. Insufficient funds.")
    print(f"Available Balance: R{balance:.2f}")

print("\nThank you for using FNB ATM!")
```
