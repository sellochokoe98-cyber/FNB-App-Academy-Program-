```python
# Multi-Function Calculator

print("=== Multi-Function Calculator ===")

# 1. COLLECT INPUT SAFELY with float() and type casting
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# 2. BASIC OPERATIONS
addition = round(num1 + num2, 2)
subtraction = round(num1 - num2, 2)
multiplication = round(num1 * num2, 2)

# 3. ADVANCED OPERATIONS
floor_div = round(num1 // num2, 2) if num2 != 0 else "Error"
modulus = round(num1 % num2, 2) if num2 != 0 else "Error"

# 4. HANDLE DIVISION BY ZERO
if num2 == 0:
    division = "Error: Cannot divide by zero"
else:
    division = round(num1 / num2, 2)

# 5. DISPLAY RESULTS IN A FORMATTED TABLE using f-strings
print("\n--- Results ---")
print(f"| {'Operation':<15} | {'Result':<20} |")
print(f"|{'-'*17}|{'-'*22}|")
print(f"| {'Addition':<15} | {addition:<20} |")
print(f"| {'Subtraction':<15} | {subtraction:<20} |")
print(f"| {'Multiplication':<15} | {multiplication:<20} |")
print(f"| {'Division':<15} | {division:<20} |")
print(f"| {'Floor Division':<15} | {floor_div:<20} |")
print(f"| {'Modulus':<15} | {modulus:<20} |")
```
