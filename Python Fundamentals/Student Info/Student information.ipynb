```python
# Student Info Formatter

first_name = input("Enter your first name: ")
surname = input("Enter your surname: ")
age = int(input("Enter your age: "))
fav_number = float(input("Enter your favourite number: "))

full_name = f"{first_name} {surname}"
name_upper = full_name.upper()
name_title = full_name.title()

age_in_months = age * 12
fav_number_rounded = round(fav_number, 2)

print("\n" + "=" * 35)
print(f"Welcome, {full_name}!")
print("=" * 35)
print(f"Name UPPERCASE: {name_upper}")
print(f"Name Title Case: {name_title}")
print(f"Age: {age} years")
print(f"Age in months: {age_in_months} months")
print(f"Favourite number: {fav_number_rounded}")
print("=" * 35)

print("\nData Types:")
print(f"First name: {type(first_name)}")
print(f"Surname: {type(surname)}")
print(f"Age: {type(age)}")
print(f"Favourite number: {type(fav_number)}")
```
