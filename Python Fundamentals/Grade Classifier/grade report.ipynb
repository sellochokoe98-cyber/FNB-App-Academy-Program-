```python
# grade_report.py

students = [
    {"name": "John", "maths": 80, "english": 70, "science": 90},
    {"name": "Mary", "maths": 60, "english": 65, "science": 70},
    {"name": "Peter", "maths": 50, "english": 55, "science": 60},
    {"name": "Sarah", "maths": 40, "english": 45, "science": 50},
    {"name": "David", "maths": 90, "english": 85, "science": 95}
]

results = []
total = 0
highest = 0
lowest = 100

for s in students:
    avg = (s["maths"] + s["english"] + s["science"]) / 3

    if avg >= 75:
        grade = "A"
        status = "Pass"
    elif avg >= 50:
        grade = "C"
        status = "Pass"
    else:
        grade = "F"
        status = "Fail"

    results.append({"name": s["name"], "average": avg, "grade": grade, "status": status})
    total += avg
    if avg > highest:
        highest = avg
    if avg < lowest:
        lowest = avg

print("Class Report")
for r in results:
    print(r)

print("Class Average:", total / len(students))
print("Highest:", highest)
print("Lowest:", lowest)

while True:
    name = input("Search student (or quit): ")
    if name.lower() == "quit":
        break

    for r in results:
        if r["name"].lower() == name.lower():
            print(r)
            break
```
