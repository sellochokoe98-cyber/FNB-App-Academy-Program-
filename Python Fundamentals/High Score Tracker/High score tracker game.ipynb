```python
# The High-Score Tracker Game

print("=== High-Score Tracker Game ===")

# Start an intentional infinite loop
while True:
    # Ask the user for a game score
    score = input("Enter your game score (or type 'stop' to quit): ").strip()

    # Check if the user wants to stop
    if score.lower() == "stop":
        print("Game session ended!")
        break

    # Convert the input to an integer
    score = int(score)

    # Check if the score is greater than 100
    if score > 100:
        print("Wow! That's a new high score!")
    else:
        print("Good try, keep playing!")
```
